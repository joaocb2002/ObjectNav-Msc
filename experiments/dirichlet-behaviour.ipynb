```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import dirichlet

# Define a mixed-alpha Dirichlet case
alpha_mixed = [0.99, 0.99, 0.95]  # Parameters for the Dirichlet distribution

# Create a grid of points over the 2-simplex (triangle)
n = 200
x = np.linspace(0.001, 0.999, n)
y = np.linspace(0.001, 0.999, n)
X, Y = np.meshgrid(x, y)
Z = np.zeros_like(X)

for i in range(n):
    for j in range(n):
        a = X[i, j]
        b = Y[i, j]
        if a + b < 1:
            c = 1 - a - b
            Z[i, j] = dirichlet.pdf([a, b, c], alpha_mixed)
        else:
            Z[i, j] = np.nan

# Plot the density over the 2-simplex
plt.figure(figsize=(7, 6))
plt.contourf(X, Y, Z, levels=100, cmap='plasma')
plt.title(f'Dirichlet PDF (3-category) with α={alpha_mixed}')
plt.xlabel('x1')
plt.ylabel('x2')
plt.colorbar(label='Density')
plt.grid(True)
plt.show()
```
