```python
def compute_dirichlet_alpha_i(target_prob: float, num_categories: int, other_alpha: float = 1.0) -> float:
    """
    Compute the alpha_i value for the Dirichlet distribution such that the expected
    value of category i is equal to target_prob, given all other alphas are set to other_alpha.
    
    Parameters:
    - target_prob: Desired expected probability for the i-th component (e.g. 0.85)
    - num_categories: Total number of categories (e.g. 80)
    - other_alpha: Value for all other alpha parameters (default is 1.0)
    
    Returns:
    - Required alpha_i value
    """
    if not (0 < target_prob < 1):
        raise ValueError("target_prob must be between 0 and 1 (exclusive).")
    if num_categories < 2:
        raise ValueError("num_categories must be at least 2.")

    sum_other_alphas = other_alpha * (num_categories - 1)
    alpha_i = (target_prob * sum_other_alphas) / (1 - target_prob)
    return alpha_i

# Example usage
target_probability = 0.3333333
num_categories = 27
alpha_i_value = compute_dirichlet_alpha_i(target_probability, num_categories)
print(f"Required alpha_i: {alpha_i_value:.4f}")

# Entropy calculation
import numpy as np
def compute_entropy(probabilities: np.ndarray) -> float:
    """
    Compute the entropy of a probability distribution.
    
    Parameters:
    - probabilities: A numpy array of probabilities (must sum to 1)
    
    Returns:
    - Entropy value
    """
    if not np.isclose(np.sum(probabilities), 1):
        raise ValueError("Probabilities must sum to 1.")
    
    return -np.sum(probabilities * np.log(probabilities))

# Compute in vector of previous example
vec = np.array([1] * 27, dtype=float)
vec[0] = alpha_i_value
vec /= np.sum(vec)
entropy_value = compute_entropy(vec)
print(f"Entropy of the distribution: {entropy_value:.4f}")
```
