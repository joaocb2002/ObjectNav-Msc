```python
import numpy as np
# Epsilon schedule
def get_epsilon(episode, epsilon_start=1.0, epsilon_end=0.01, epsilon_decay=30):
    """
    Compute the epsilon value for the given episode.
    """
    return epsilon_end + (epsilon_start - epsilon_end) * np.exp(-episode / epsilon_decay)

# Plot get_epsilon function
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0, 100, 1)
y = get_epsilon(x)
plt.plot(x, y)
plt.xlabel('Episode')
plt.ylabel('Epsilon')
plt.title('Epsilon Decay Schedule')
plt.grid()
plt.show()
```
