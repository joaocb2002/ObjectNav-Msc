```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gamma

# Parameters of the Gamma distribution
a = 10  # shape parameter, must be > 1
b = 10  # rate parameter
scale = 1 / b  # scale is the inverse of rate in scipy's gamma

# Generate x values
x = np.linspace(0, 10, 1000)
# Compute the PDF of the Gamma distribution
pdf = gamma.pdf(x, a, scale=scale)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, pdf, label=f'Gamma(a={a}, b={b})', lw=2)
plt.title('Gamma Distribution PDF')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
