```python
import numpy as np
import matplotlib.pyplot as plt
FN_RATE = 0.3283 # Learned from calibration data

# Parameters
lambd = 1.0

# X values
x = np.linspace(0, 10, 1000)

# Manual computation of PDF
pdf = lambd * np.exp(-lambd * x)*(1-FN_RATE)

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(x, pdf, label=f'Manual Exponential PDF (λ={lambd})')
plt.title('Exponential Distribution (Manual Calculation)')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.grid(True)
plt.legend()
plt.show()
```
