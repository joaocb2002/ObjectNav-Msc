```python
import json
import pandas as pd
from tabulate import tabulate

# Load the JSON data
with open('metrics.json', 'r') as f:
    data = json.load(f)

# Group by simulation index
simulations = {}
for entry in data:
    sim_index = entry["simulation_index"]
    if sim_index not in simulations:
        simulations[sim_index] = {
            "meta": {
                "scene": entry["scene"],
                "target_object": entry["target_object"],
                "confidence_threshold": entry["confidence_threshold"],
                "max_iter_coefficient": entry["max_iter_coefficient"],
                "location_error_threshold": entry["location_error_threshold"],
                "pseudo_count_threshold": entry["pseudo_count_threshold"]
            },
            "entries": []
        }
    simulations[sim_index]["entries"].append(entry)

# Define the desired method order
method_order = ["RANDOM", "BASELINE", "DIRICHLET"]

# Print summary tables
for sim_index, sim_data in simulations.items():
    meta = sim_data["meta"]
    entries = sim_data["entries"]

    print(f"\n\n\nSimulation Index: {sim_index}")
    print(f"Scene: {meta['scene']}")
    print(f"Target Object: {meta['target_object']}")
    print(f"Confidence Threshold: {meta['confidence_threshold']}")
    print(f"Max Iter Coefficient: {meta['max_iter_coefficient']}")
    print(f"Pseudo Count Threshold: {meta['pseudo_count_threshold']}")
    print(f"Location Error Threshold: {meta['location_error_threshold']}\n")

    df = pd.DataFrame(entries)
    df = df[[
        "search_method",
        "success_rate",
        "total_avg_num_actions",
        "total_avg_tavelled_distance",
        "success_avg_num_actions",
        "success_avg_travelled_distance",
        "success_avg_location_error"
    ]]

    df = df.rename(columns={
        "search_method": "Method",
        "success_rate": "Success Rate (%)",
        "total_avg_num_actions": "Total Avg Actions",
        "total_avg_tavelled_distance": "Total Avg Distance",
        "success_avg_num_actions": "Success Avg Actions",
        "success_avg_travelled_distance": "Success Avg Distance",
        "success_avg_location_error": "Avg Location Error"
    })

    # Set method order as categorical and sort
    df["Method"] = pd.Categorical(df["Method"], categories=method_order, ordered=True)
    df = df.sort_values("Method")

    print(tabulate(df, headers="keys", tablefmt="grid", showindex=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
methods = ['RANDOM', 'BASELINE', 'DIRICHLET']
total_avg_actions = [1067.26, 203.385, 137.545]
total_avg_distance = [86.0144, 42.1084, 29.2143]
success_rate = [27.5, 100, 100]

# Bar width and position
x = np.arange(len(methods))
width = 0.25

# Create subplots
fig, axs = plt.subplots(1, 3, figsize=(12, 4))

# Chart 1: Total Avg Actions
axs[0].bar(x, total_avg_actions, width, color='skyblue')
axs[0].set_title('Total Avg Actions')
axs[0].set_ylabel('Actions')
axs[0].set_xticks(x)
axs[0].set_xticklabels(methods)
axs[0].grid(True, axis='y', linestyle='--', alpha=0.5)

# Chart 2: Total Avg Distance
axs[1].bar(x, total_avg_distance, width, color='lightgreen')
axs[1].set_title('Total Avg Distance')
axs[1].set_ylabel('Distance (m)')
axs[1].set_xticks(x)
axs[1].set_xticklabels(methods)
axs[1].grid(True, axis='y', linestyle='--', alpha=0.5)

# Chart 3: Success Rate
axs[2].bar(x, success_rate, width, color='salmon')
axs[2].set_title('Success Rate')
axs[2].set_ylabel('Success (%)')
axs[2].set_xticks(x)
axs[2].set_xticklabels(methods)
axs[2].set_ylim(0, 110)
axs[2].grid(True, axis='y', linestyle='--', alpha=0.5)

# Tight layout
fig.suptitle('Performance Comparison Across Methods', fontsize=14, weight='bold')
fig.tight_layout(rect=[0, 0.03, 1, 0.90])

# Save or show
plt.savefig('performance_comparison.png', dpi=300)
plt.show()
```
