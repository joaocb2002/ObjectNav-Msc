```python
import json
import pandas as pd
from tabulate import tabulate

# Load the JSON data
with open('metrics.json', 'r') as f:
    data = json.load(f)

# Group by simulation index
simulations = {}
for entry in data:
    sim_index = entry["simulation_index"]
    if sim_index not in simulations:
        simulations[sim_index] = {
            "meta": {
                "scene": entry["scene"],
                "target_object": entry["target_object"],
                "confidence_threshold": entry["confidence_threshold"],
                "max_iter_coefficient": entry["max_iter_coefficient"],
                "location_error_threshold": entry["location_error_threshold"],
            },
            "entries": []
        }
    simulations[sim_index]["entries"].append(entry)

# Define the desired method order
method_order = ["RANDOM", "BASELINE", "DIRICHLET"]

# Print summary tables
for sim_index, sim_data in simulations.items():
    meta = sim_data["meta"]
    entries = sim_data["entries"]

    print(f"\n\n\nSimulation Index: {sim_index}")
    print(f"Scene: {meta['scene']}")
    print(f"Target Object: {meta['target_object']}")
    print(f"Confidence Threshold: {meta['confidence_threshold']}")
    print(f"Max Iter Coefficient: {meta['max_iter_coefficient']}")
    print(f"Location Error Threshold: {meta['location_error_threshold']}\n")

    df = pd.DataFrame(entries)
    df = df[[
        "search_method",
        "success_rate",
        "total_avg_num_actions",
        "total_avg_tavelled_distance",
        "success_avg_num_actions",
        "success_avg_travelled_distance",
        "success_avg_location_error"
    ]]

    df = df.rename(columns={
        "search_method": "Method",
        "success_rate": "Success Rate (%)",
        "total_avg_num_actions": "Total Avg Actions",
        "total_avg_tavelled_distance": "Total Avg Distance",
        "success_avg_num_actions": "Success Avg Actions",
        "success_avg_travelled_distance": "Success Avg Distance",
        "success_avg_location_error": "Avg Location Error"
    })

    # Set method order as categorical and sort
    df["Method"] = pd.Categorical(df["Method"], categories=method_order, ordered=True)
    df = df.sort_values("Method")

    print(tabulate(df, headers="keys", tablefmt="grid", showindex=False))
```
