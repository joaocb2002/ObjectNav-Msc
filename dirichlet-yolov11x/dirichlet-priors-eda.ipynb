```python
# Load Dirichlet priors
with open('dirichlet_priors.pkl', 'rb') as f:
    dirichlet_priors = pickle.load(f)

# Load bins per class
with open('out_bins_per_class.json', 'r') as f:
    classes_bins = json.load(f)

# Print the number of priors for each key
for key in dirichlet_priors:
    print(f"\n\n {key}: {len(dirichlet_priors[key])} priors")

    for i in range(10):
        if dirichlet_priors[key][i] is None:
            print(f"Prior {i} is None", end=' ')
        else:
        
            print(np.argmax(dirichlet_priors[key][i]), end=' ')
            print(max(dirichlet_priors[key][i]), end=' ')
        
        print(f"Bins: {classes_bins[key][i]}")
```
