```python
import pickle
import numpy as np
import dirichlet
import matplotlib.pyplot as plt
import seaborn as sns
import time
```

```python
# Load the pickle file
with open("vector_scores_struct.pkl", "rb") as f:
    vector_scores_struct = pickle.load(f)
```

```python
# Store alphas: same structure as input
dirichlet_alphas = {}

for cls, bins in vector_scores_struct.items():
    dirichlet_alphas[cls] = {}
    for bin_idx, vectors in bins.items():
        start = time.time()
        if len(vectors) > 1:
            data = np.array(vectors)
            try:
                alpha = dirichlet.mle(data, method='fixedpoint')  # or 'fixed_point'
                dirichlet_alphas[cls][bin_idx] = alpha
            except Exception as e:
                print(f"Failed to fit Dirichlet for {cls} bin {bin_idx}: {e}")
                dirichlet_alphas[cls][bin_idx] = None
        else:
            dirichlet_alphas[cls][bin_idx] = None  # not enough data
        
        print(f"{cls} bin {bin_idx} done for {len(vectors)} vectors.")
```

```python
with open("dirichlet_priors.pkl", "wb") as f:
    pickle.dump(dirichlet_alphas, f)
```
