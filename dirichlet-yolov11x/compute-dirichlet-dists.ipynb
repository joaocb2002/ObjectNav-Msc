```python
import pickle
import numpy as np
import dirichlet
from mylib import dirichlet_custom
import matplotlib.pyplot as plt
import seaborn as sns
import time
```

```python
# Load the pickle file
with open("vector-scores-struct.pkl", "rb") as f:
    vector_scores_struct = pickle.load(f)
```

```python
# Store alphas: same structure as input
dirichlet_alphas = {}

for cls, bins in vector_scores_struct.items():
    dirichlet_alphas[cls] = {}

    # print(f"\n\n ******************* Processing class: {cls} *******************")
    for bin_idx, vectors in bins.items():

        # print(f"\n\nProcessing bin {bin_idx} with {len(vectors)} vectors...")
        start = time.time()
        if len(vectors) > 1:
            data = np.array(vectors)
            try:
                # MAP estimation
                alpha_map = dirichlet_custom.dirichlet_map(data, a_prior=10.0, b_prior=10.0)
                dirichlet_alphas[cls][bin_idx] = alpha_map
                # print("\nMAP alpha:", alpha_map)
                # print(f"Argmax: {np.argmax(alpha_map)} \tMax: {np.max(alpha_map)} \tArgmin: {np.argmin(alpha_map)} \tMin: {np.min(alpha_map)} \tSum: {np.sum(alpha_map)} \tMean: {np.mean(alpha_map)}")

                # # MLE estimation
                # alpha_mle = dirichlet_custom.dirichlet_mle(data)
                # print("\nMLE alpha:", alpha_mle)
                # print(f"Argmax: {np.argmax(alpha_mle)} \tMax: {np.max(alpha_mle)} \tArgmin: {np.argmin(alpha_mle)} \tMin: {np.min(alpha_mle)} \tSum: {np.sum(alpha_mle)} \tMean: {np.mean(alpha_mle)}")

                # # Minka library MLE estimation
                # alpha_minka = dirichlet.mle(data, method='fixedpoint')
                # print("\nMinka MLE alpha:", alpha_minka)
                # print(f"Argmax: {np.argmax(alpha_minka)} \tMax: {np.max(alpha_minka)} \tArgmin: {np.argmin(alpha_minka)} \tMin: {np.min(alpha_minka)} \tSum: {np.sum(alpha_minka)} \tMean: {np.mean(alpha_minka)}")

            except Exception as e:
                print(f"Failed to fit Dirichlet for {cls} bin {bin_idx}: {e}")
                dirichlet_alphas[cls][bin_idx] = None
        else:
            dirichlet_alphas[cls][bin_idx] = None  # not enough data
        
        print(f"{cls} bin {bin_idx} done for {len(vectors)} vectors.")
```

```python
# Print the alphas for each class and bin
for cls, bins in dirichlet_alphas.items():
    print(f"\nClass: {cls}")
    for bin_idx, alpha in bins.items():
        if alpha is not None:
            print(f"\nBin {bin_idx}: {alpha}")
            print(f"Argmax: {np.argmax(alpha)} \tMax: {np.max(alpha)} \tArgmin: {np.argmin(alpha)} \tMin: {np.min(alpha)} \tSum: {np.sum(alpha)} \tMean: {np.mean(alpha)}")
            print(f"Expected value: {alpha/np.sum(alpha)}")
            print(f"Argmax expected value: {np.argmax(alpha/np.sum(alpha))} \tMax expected value: {np.max(alpha/np.sum(alpha))} \tArgmin expected value: {np.argmin(alpha/np.sum(alpha))} \tMin expected value: {np.min(alpha/np.sum(alpha))} \tSum expected value: {np.sum(alpha/np.sum(alpha))} \tMean expected value: {np.mean(alpha/np.sum(alpha))}")
        else:
            print(f"Bin {bin_idx}: Not enough data to estimate alpha.")
```
